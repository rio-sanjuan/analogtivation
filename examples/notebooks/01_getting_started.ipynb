````python
import numpy as np
import matplotlib.pyplot as plt
from analogtivation.core import (
    clock_activation,
    wave_activation,
    fourier_activation,
    quantum_activation,
    lorenz_activation,
    logistic_map_activation,
    mandelbrot_activation
)

# Set random seed for reproducibility
np.random.seed(42)
````

## Visualizing Activation Functions

Let's visualize how different activation functions transform inputs.

````python
# Create input range
x = np.linspace(-3, 3, 1000)

# Create subplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

# Plot different activations
activations = [
    ('Clock', clock_activation(x)),
    ('Wave', wave_activation(x, frequencies=[1.0, 2.0], amplitudes=[0.5, 0.3])),
    ('Fourier', fourier_activation(x, n_terms=5)),
    ('Quantum', quantum_activation(x, n_states=3)),
    ('Lorenz', lorenz_activation(x)),
    ('Logistic Map', logistic_map_activation(x, r=3.7, iterations=5))
]

for idx, (name, y) in enumerate(activations):
    axes[idx].plot(x, y, 'b-', linewidth=2)
    axes[idx].plot(x, x, 'k--', alpha=0.3, label='y=x')
    axes[idx].set_title(f'{name} Activation', fontsize=14)
    axes[idx].set_xlabel('Input')
    axes[idx].set_ylabel('Output')
    axes[idx].grid(True, alpha=0.3)
    axes[idx].legend()

plt.tight_layout()
plt.show()
````

## Time-Based Activations

Time-based activations change their behavior based on the current time. This can be useful for:
- Modeling temporal patterns
- Creating adaptive networks that change over time
- Incorporating real-world time dependencies

````python
# Demonstrate how clock activation changes over time
from time import sleep
import datetime

x_sample = np.array([1.0, -0.5, 2.0, -1.5])

print("Clock activation values at different times:")
for i in range(3):
    current_time = datetime.datetime.now().strftime("%H:%M:%S")
    y = clock_activation(x_sample)
    print(f"Time: {current_time}, Output: {y}")
    sleep(2)  # Wait 2 seconds
````

## Wave-Based Activations

Wave-based activations use periodic functions to create complex patterns.

````python
# Explore different wave configurations
x = np.linspace(-5, 5, 1000)

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

# Different frequency and amplitude combinations
configs = [
    ([1.0], [1.0], "Single wave"),
    ([1.0, 2.0], [1.0, 0.5], "Two harmonics"),
    ([1.0, 2.0, 3.0], [1.0, 0.5, 0.33], "Three harmonics"),
    ([0.5, 1.5, 2.5, 3.5], [1.0, 0.7, 0.5, 0.3], "Complex pattern")
]

for idx, (freqs, amps, title) in enumerate(configs):
    y = wave_activation(x, frequencies=freqs, amplitudes=amps)
    axes[idx].plot(x, y, 'b-', linewidth=2)
    axes[idx].set_title(title)
    axes[idx].set_xlabel('Input')
    axes[idx].set_ylabel('Output')
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Chaos-Based Activations

Chaos-based activations introduce controlled randomness and complex dynamics.

````python
# Explore chaos parameters
x = np.linspace(-2, 2, 500)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Lorenz with different parameters
for sigma in [5.0, 10.0, 15.0]:
    y = lorenz_activation(x, sigma=sigma)
    axes[0].plot(x, y, label=f'σ={sigma}')
axes[0].set_title('Lorenz Activation')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Logistic map with different r values
for r in [2.5, 3.5, 3.9]:
    y = logistic_map_activation(x, r=r, iterations=5)
    axes[1].plot(x, y, label=f'r={r}')
axes[1].set_title('Logistic Map Activation')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

# Mandelbrot with different iterations
for max_iter in [2, 5, 10]:
    y = mandelbrot_activation(x, max_iter=max_iter)
    axes[2].plot(x, y, label=f'iter={max_iter}')
axes[2].set_title('Mandelbrot Activation')
axes[2].legend()
axes[2].grid(True, alpha=0.3)

for ax in axes:
    ax.set_xlabel('Input')
    ax.set_ylabel('Output')

plt.tight_layout()
plt.show()
````

## Quantum Activation

The quantum activation function maps inputs to discrete quantum states.

````python
# Visualize quantum state mapping
x = np.linspace(-3, 3, 1000)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Different number of states
for idx, n_states in enumerate([2, 4, 8]):
    y = quantum_activation(x, n_states=n_states)
    axes[idx].plot(x, y, 'b-', linewidth=2)
    axes[idx].set_title(f'Quantum Activation (n_states={n_states})')
    axes[idx].set_xlabel('Input')
    axes[idx].set_ylabel('State Index')
    axes[idx].grid(True, alpha=0.3)
    axes[idx].set_ylim(-0.5, n_states - 0.5)
    axes[idx].set_yticks(range(n_states))

plt.tight_layout()
plt.show()
````

## Performance Comparison

Let's compare the computational performance of different activations.

````python
import time

# Test data
test_sizes = [1000, 10000, 100000]
activations = [
    ('Clock', lambda x: clock_activation(x)),
    ('Wave', lambda x: wave_activation(x)),
    ('Fourier', lambda x: fourier_activation(x, n_terms=5)),
    ('Quantum', lambda x: quantum_activation(x, n_states=4)),
    ('Lorenz', lambda x: lorenz_activation(x)),
    ('Logistic', lambda x: logistic_map_activation(x, iterations=5))
]

# Benchmark
results = {name: [] for name, _ in activations}

for size in test_sizes:
    x = np.random.randn(size).astype(np.float32)
    
    for name, func in activations:
        start = time.time()
        _ = func(x)
        elapsed = time.time() - start
        results[name].append(elapsed * 1000)  # Convert to milliseconds

# Plot results
plt.figure(figsize=(10, 6))
for name in results:
    plt.plot(test_sizes, results[name], 'o-', label=name, linewidth=2, markersize=8)

plt.xlabel('Input Size')
plt.ylabel('Time (ms)')
plt.title('Activation Function Performance')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xscale('log')
plt.yscale('log')
plt.show()
````
